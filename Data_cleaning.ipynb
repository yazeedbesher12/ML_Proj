#### Handling Missing Values

The column `TotalCharges` was originally stored as an object, although it should be numerical. It was converted to a numeric type using `pd.to_numeric()`.

During the conversion, blank or non-numeric values were converted into missing values. These missing values were found in customers with `tenure = 0`, which means they are new customers and have not accumulated total charges yet.

Therefore, the missing values in `TotalCharges` were replaced with 0.

```python
df = df.drop('customerID', axis=1)
df.head()
```

```python
df = df.drop(columns=['PhoneService', 'MultipleLines', 'DeviceProtection', 'OnlineBackup', 'gender'])
```

#### Removing Low-Correlation Features

Based on the correlation analysis with the target variable `Churn`, some features showed very weak relationships with the target. These features were removed to reduce noise and simplify the dataset.

The removed features were:
- gender
- PhoneService
- MultipleLines
- DeviceProtection
- OnlineBackup

These features had low correlation values with `Churn`, so they were considered less useful for prediction compared to stronger features such as tenure, contract type, internet service, payment method, monthly charges, and total charges.

```python
categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:
    print(col)
    print(df[col].unique())
    print('-' * 40)
```

### Checking Categorical Inconsistencies

The unique values of categorical columns were checked to identify possible inconsistencies such as spelling differences, duplicated categories, or unexpected values.

Values such as `No internet service` and `No phone service` were kept because they represent meaningful categories, not missing values.

```python
df['Churn'] = df['Churn'].map({'No': 0, 'Yes': 1})
```

#### Encoding Target Variable

The target variable `Churn` was converted into numerical format:

- No = 0
- Yes = 1

This conversion is necessary because machine learning models require numerical input.

```python
categorical_cols = df.select_dtypes(include='object').columns

df_encoded = pd.get_dummies(df, columns=categorical_cols, drop_first=True)

bool_cols = df_encoded.select_dtypes(include='bool').columns

df_encoded[bool_cols] = df_encoded[bool_cols].astype(int)
df_encoded.head()
```

#### Encoding Categorical Variables

Categorical variables were converted into numerical format using one-hot encoding.

One-hot encoding creates binary columns with values 0 and 1. The parameter `drop_first=True` was used to avoid redundant columns and reduce multicollinearity.

For example, a categorical feature with three categories is converted into two binary columns, where the missing category acts as the reference category.

```python
df_encoded.shape
```

```python
num_cols = ['tenure', 'MonthlyCharges', 'TotalCharges']
df_encoded[num_cols].describe()
```

The numerical features have different ranges. For example, `TotalCharges` has a much larger range than `tenure` and `MonthlyCharges`. Therefore, scaling may be useful depending on the machine learning algorithm used.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))
sns.boxplot(data=df_encoded[num_cols])
plt.title('Numerical Features Before Scaling')
plt.show()
```

So We Should Scaling !! 

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

num_cols = ['tenure', 'MonthlyCharges', 'TotalCharges']

df_encoded[num_cols] = scaler.fit_transform(df_encoded[num_cols])
```

```python
df_encoded[['tenure', 'MonthlyCharges', 'TotalCharges']].describe()
```

#### Scaling Numerical Features

The numerical features `tenure`, `MonthlyCharges`, and `TotalCharges` were scaled using StandardScaler.

StandardScaler was used because these features have different ranges. For example, `TotalCharges` has a much larger range than `tenure` and `MonthlyCharges`.

Scaling helps ensure that numerical features contribute more equally to machine learning models, especially algorithms that are sensitive to feature scale such as Logistic Regression, KNN, and SVM.
