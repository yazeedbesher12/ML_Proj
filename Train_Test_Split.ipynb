# Train/Test Split

We load the cleaned dataset and split it into:

- 70% training data
- 30% testing data

The split uses `stratify=y` so the churn distribution stays similar in both training and testing sets.  
reference : [scikit-learn train_test_split documentation](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.train_test_split.html)



```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

## 1. Load the cleaned dataset

```python
df = pd.read_csv("cleaned.csv")

df.head()
```

```python
df.shape
```

## 2. Check the target column

The target variable is `Churn`.

`0` means the customer did not churn.

`1` means the customer churned.


```python
df["Churn"].value_counts()
```

```python
df["Churn"].value_counts(normalize=True)
```

## 3. Separate features and target

`X` contains all input features.

`y` contains the target variable `Churn`.


```python
X = df.drop("Churn", axis=1)
y = df["Churn"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

## 4. Split the dataset

We split the dataset into:

- 70% training
- 30% testing

`stratify=y` preserves the churn ratio in both sets.


```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    train_size=0.70,
    random_state=42,
    stratify=y
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

## 5. Verify churn distribution after splitting

The distribution should be almost the same in the original dataset, training set, and testing set.


```python
print("Original churn distribution:")
print(y.value_counts(normalize=True))

print("\nTraining churn distribution:")
print(y_train.value_counts(normalize=True))

print("\nTesting churn distribution:")
print(y_test.value_counts(normalize=True))
```

## 6. Save the split datasets

These files can be used in the next notebooks:

- `X_train.csv`
- `X_test.csv`
- `y_train.csv`
- `y_test.csv`

training and testing sets are placed into their own folders


```python
import os

os.makedirs("train", exist_ok=True)
os.makedirs("test", exist_ok=True)
X_train.to_csv("train/X_train.csv", index=False)
X_test.to_csv("test/X_test.csv", index=False)
y_train.to_csv("train/y_train.csv", index=False)
y_test.to_csv("test/y_test.csv", index=False)

print("Train/test split files saved successfully.")
```
