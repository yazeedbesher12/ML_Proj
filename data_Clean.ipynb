### Correlation Analysis with Churn

To identify the most important variables related to customer churn, the target variable `Churn` was converted into numerical format:

- No = 0
- Yes = 1

```python
df_corr = df.copy()
```

```python
df_corr['Churn'] = df_corr['Churn'].map({'No': 0, 'Yes': 1})
```

```python
df_corr['TotalCharges'] = pd.to_numeric(df_corr['TotalCharges'], errors='coerce')
```

```python
numeric_corr = df_corr.corr(numeric_only=True)['Churn'].sort_values(ascending=False)

numeric_corr
```

#### We Notice That the Numeric Variables does not affect the `Churn` so much , amd 'TotalCharges , tenue' has a negative effect on the `Churn` . 
So i well Go To check the categorial variables . 

```python
df_encoded = df_corr.copy()

df_encoded = pd.get_dummies(df_encoded, drop_first=True)

corr_with_churn = df_encoded.corr()['Churn'].sort_values(ascending=False)

corr_with_churn
```
